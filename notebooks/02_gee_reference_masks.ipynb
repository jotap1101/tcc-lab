Rasteriza (idempotente) as glebas de café do Geoportal do Café (Emater-MG, mapeamento de 2018 validado em campo) a partir dos GeoJSONs por município de `data/external/emater/`: reprojeção para EPSG:31983, reparo de geometrias (buffer(0)), clipe ao AOI e rasterização no grid de 10 m. ## Calibração do limiar — AlphaEarth (modelo 2025b)

Estima a área de café do AOI para vários limiares de probabilidade do modelo 2025b e compara com as fontes de referência, orientando a escolha do `probability_threshold` a registrar em `src/config.yaml`.

```python
# Estima a área de café por limiar da AlphaEarth 2025b e compara com as referências.
import ee

from src.data.mask_utils import reference_year

alphaearth_cfg = config["ground_truth"]["sources"]["alphaearth"]
alphaearth_year = reference_year("alphaearth")
probability = (
    ee.ImageCollection(alphaearth_cfg["collection_id"])
    .filterDate(f"{alphaearth_year}-01-01", f"{alphaearth_year}-12-31")
    .mosaic()
    .select("probability")
)
scale = int(config["data"]["export"]["scale"])
aoi_area_km2 = float(aoi_row["AREA_KM2"])

threshold_areas: dict[str, float] = {}
for threshold in [0.20, 0.25, 0.28, 0.30, 0.35, 0.40, 0.50]:
    share = (
        probability.gte(threshold)
        .clip(aoi_geometry)
        .reduceRegion(ee.Reducer.mean(), aoi_geometry, scale, bestEffort=True)
        .get("probability")
        .getInfo()
    )
    threshold_areas[f"{threshold:.2f}"] = round(share * aoi_area_km2, 2)

print("Área de café estimada por limiar (AlphaEarth 2025b):")
for threshold_label, area in threshold_areas.items():
    print(f"  limiar {threshold_label}: {area} km²")
print("Referências: MapBiomas e Emater (veja as áreas das máscaras deste estágio).")
print("Registre o limiar escolhido em config.yaml (probability_threshold) e reexecute")
print("a célula da máscara da AlphaEarth — o nome do GeoTIFF embute o limiar.")
```
