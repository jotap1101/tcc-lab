## Treinamento por dobra

Executa o protocolo de treino único (`src.trainer.train_fold`) para cada dobra da validação espacial da fonte escolhida, construindo um U-Net novo por dobra com a semente da dobra (`seed + fold`) fixada antes da inicialização dos pesos. Artefatos de uma execução anterior só são reutilizados quando o fingerprint do dataset e o hash da configuração coincidem; caso contrário, a dobra é retreinada automaticamente. A célula exige GPU: interrompe a execução quando o acelerador não está ativo, evitando um treino inviável em CPU.

```python
# Executa o protocolo de treino único por dobra da fonte escolhida (reutiliza artefatos).
import torch

from src.data.spatial_split import fold_count
from src.models.unet import build_unet
from src.trainer import train_fold
from src.utils import set_all_seeds

# Exige GPU: o treino em CPU é inviável para 512x512 por 100 épocas.
if not torch.cuda.is_available():
    raise RuntimeError(
        "GPU indisponível: ative o acelerador de GPU (T4/P100) na plataforma antes de treinar."
    )
device = torch.device("cuda")
props = torch.cuda.get_device_properties(0)
print(f"Dispositivo: {device} | {torch.cuda.get_device_name(0)} | {props.total_memory / 1e9:.1f} GB")
results = []
for fold in range(fold_count()):
    # Fixa a semente antes de inicializar os pesos, tornando cada dobra reproduzível.
    set_all_seeds(ctx.seed + fold)
    unet = build_unet()
    result = train_fold(unet, "unet", storage_paths, fold, device, source_name=source_name)
    results.append(result)
```

## Resumo do treinamento e artefatos

Exibe o resumo por dobra (melhor época e IoU de validação), persiste o metadata de execução (configuração, fonte, fingerprint do manifesto e versões do ambiente) em `MyDrive/tcc/artifacts/runs/unet/<fonte>/`.

```python
# Exibe o resumo do treinamento e persiste o metadata de execução da fonte.
from src.trainer import save_run_metadata
from src.utils import print_summary

meta_file = save_run_metadata(storage_paths, "unet", source_name)
summary = {"Modelo": "unet", "Fonte de treinamento": source_name, "Dobras treinadas": len(results)}
for result in results:
    tag = " (reutilizada)" if result.skipped else ""
    summary[f"Dobra {result.fold} — IoU val"] = (
        f"{result.val_metrics['iou']:.4f} (época {result.best_epoch}){tag}"
    )
summary["Pesos por dobra"] = str(results[0].weights_path.parent)
summary["Metadata de execução"] = str(meta_file)
print_summary(summary, "09")
```
