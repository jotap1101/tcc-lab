# Aquisição da imagem Sentinel-2 (estágio 01)

Carrega a malha vetorial do IBGE, define o polígono da Região Geográfica Imediata de Guaxupé (310044) como área de estudo (AOI) e adquire o mosaico livre de nuvens do Sentinel-2 Level-2A (bandas B2, B3, B4 e B8) via Google Earth Engine, exportando o resultado em GeoTIFF para o armazenamento canônico em `MyDrive/tcc/`.

## Bootstrap do workspace

O primeiro passo baixa e executa `src/bootstrap.py` (somente stdlib) — necessário porque o `src/` ainda não está disponível para import em uma sessão nova. O bootstrap obtém o repositório público, extrai `src/`, `data/external/` e `requirements-runtime.txt` para o workspace e adiciona o workspace ao `sys.path`. O `reload` garante que reexecuções usem a versão mais recente baixada.

```python
# Baixa e executa o bootstrap do workspace (etapa prévia ao import de src/).
import importlib
import pathlib
import sys
import urllib.request

BOOTSTRAP_URL = "https://raw.githubusercontent.com/jotap1101/tcc-lab/main/src/bootstrap.py"
pathlib.Path("bootstrap.py").write_bytes(urllib.request.urlopen(BOOTSTRAP_URL).read())
sys.path.insert(0, str(pathlib.Path.cwd()))

# Recarrega o módulo para não reutilizar uma versão antiga em cache no kernel.
bootstrap = importlib.import_module("bootstrap")
importlib.reload(bootstrap)

workspace = bootstrap.bootstrap_workspace()
print(f"Workspace: {workspace}")
```

## Área de estudo (AOI) a partir da malha IBGE

Lê a malha vetorial das Regiões Geográficas Imediatas de Minas Gerais (dado de referência versionado em `data/external/`), filtra a região `310044` (Guaxupé), reprojeta para EPSG:4326 e converte o polígono em geometria do Earth Engine.

```python
# Carrega a malha IBGE, filtra a RGI 310044 e converte o AOI para geometria do GEE.
from src.data import gee_client

mesh_path = workspace / config["aoi"]["mesh_path"]
aoi_geometry, aoi_gdf = gee_client.load_aoi_geometry(
    mesh_path, config["aoi"]["region_code"]
)
aoi_row = aoi_gdf.iloc[0]
print(f"AOI: {aoi_row['NM_RGI']} (CD_RGI={aoi_row['CD_RGI']})")
print(f"Área: {float(aoi_row['AREA_KM2']):,.1f} km²")
print(f"Extensão (EPSG:4326): {list(aoi_gdf.total_bounds)}")
```

## Anos de mosaico necessários

Lista os anos de dados requeridos pelo pipeline, derivados dos `year` próprios das fontes de ground truth habilitadas em `src/config.yaml` (ex.: MapBiomas 2025, AlphaEarth 2024, Emater 2018). Cada ano exige um mosaico Sentinel-2 próprio.

```python
# Lista os anos de dados necessários (anos das fontes de ground truth habilitadas).
from src.data.mask_utils import required_data_years

years = required_data_years()
print(f"Anos de mosaico requeridos: {', '.join(str(y) for y in years)}")
```

## Coleção Sentinel-2 filtrada por ano

Constrói, para cada ano requerido, a coleção `COPERNICUS/S2_SR_HARMONIZED` limitada ao AOI e ao período anual correspondente, descarta cenas com mais de 20% de cobertura de nuvens e mascara nuvens e sombras pela banda QA60 (bits 10 e 11), mantendo apenas as bandas B2, B3, B4 e B8.

```python
# Constrói a coleção Sentinel-2 de cada ano, filtrada por AOI, datas e nuvem.
from typing import Any

from src.data.gee_client import build_s2_collection

data_cfg = config["data"]
collections: dict[int, Any] = {}
scene_counts: dict[int, int] = {}
for year in years:
    start, end = f"{year}-01-01", f"{year}-12-31"
    collections[year] = build_s2_collection(
        aoi=aoi_geometry,
        bands=data_cfg["bands"],
        start=start,
        end=end,
        cloud_threshold=data_cfg["cloud_threshold"],
    )
    scene_counts[year] = collections[year].size().getInfo()
    print(f"{year}: {scene_counts[year]} cenas no período")
```

## Mosaico anual sem nuvens por ano

Compõe o mosaico mediano de todas as cenas de cada ano e recorta ao polígono do AOI, produzindo a imagem final com as bandas espectrais em resolução nativa de 10 m.

```python
# Compõe o mosaico mediano de cada ano e recorta ao polígono do AOI.
from src.data.gee_client import build_annual_mosaic

mosaics = {year: build_annual_mosaic(collections[year], aoi_geometry) for year in years}
print(f"Bandas do mosaico: {mosaics[years[-1]].bandNames().getInfo()}")
```

## Exportação dos mosaicos em GeoTIFF

Verifica se cada GeoTIFF já existe no caminho canônico (execuções repetidas não sobrescrevem artefatos). Como o GEE só exporta para uma pasta única na raiz do Drive (o parâmetro `folder` não aceita subcaminhos), o mosaico é exportado para a pasta `tcc` e, após a conclusão da tarefa, realocado para `MyDrive/tcc/data/raw/sentinel2/` no datum SIRGAS 2000 / UTM 23S (resolvido do `src/config.yaml`).

```python
# Garante o mosaico exportado de cada ano (reutiliza se já existir).
from src.data.gee_client import ensure_mosaic_export

export_results = {
    year: ensure_mosaic_export(mosaics[year], storage_paths, year) for year in years
}
```

## Pré-visualização dos mosaicos

Gera uma miniatura RGB de cada mosaico (bandas B4, B3, B2) via Earth Engine e salva a figura em `MyDrive/tcc/artifacts/figures/` como registro visual da etapa.

```python
# Salva a miniatura RGB de cada mosaico (idempotente).
from src.data.gee_client import save_mosaic_preview

for year in years:
    save_mosaic_preview(mosaics[year], storage_paths, year)
```

## Resumo da etapa

Exibe o resumo da aquisição: AOI definido, cenas consideradas, mosaico produzido e caminho do artefato exportado.

```python
# Exibe o resumo da etapa de aquisição.
default_year = years[-1]
target_path, exported_now = export_results[default_year]
summary = {
    "AOI": aoi_row["NM_RGI"],
    "Anos de mosaico": ", ".join(str(y) for y in years),
    f"Cenas ({default_year})": scene_counts[default_year],
    "Bandas do mosaico": mosaics[default_year].bandNames().getInfo(),
    "Mosaico exportado": str(target_path),
    "Situação": "exportado nesta execução" if exported_now else "pré-existente",
}
from src.utils import print_summary

print_summary(summary, "01")
```
